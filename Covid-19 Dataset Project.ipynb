# Data Visualization

```python
import matplotlib.pyplot as plt 
import seaborn as sns
```

# Top 10 Regions with Confirmed Cases (Bar Plot)

```python
top_10 = data.groupby('Region')['Confirmed'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_10.values, y=top_10.index, palette='viridis')
plt.title('Top 10 Regions with Maximum Confirmed Cases')
plt.xlabel('Confirmed Cases')
plt.ylabel('Region')
plt.show()
```

# Confirmed vs Deaths vs Recovered (Scatter Plot)

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=data, x='Confirmed', y='Deaths', hue='Region', legend=False)
plt.title('Confirmed vs Deaths')
plt.show()
```

# New Derivative Columns

Mortality Rate (मृत्यु दर): किस देश में संक्रमण के बाद मृत्यु की संभावना कितनी थी?

```python
data['Mortality_Rate (%)'] = (data['Deaths'] / data['Confirmed']) * 100
data.head(10)
```

Recovery Rate (ठीक होने की दर): कहाँ के मरीज़ सबसे तेज़ी से ठीक हो रहे थे?

```python
data['Recovery_Rate (%)'] = (data['Recovered'] / data['Confirmed']) * 100
data.head(10)
```
